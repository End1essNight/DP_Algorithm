```python
from T12012_vd_overshoot_multi import vd_overshoot_extraction
import os
import csv
```

```python
output_folder = r'C:\Users\FeiyuanYang\OneDrive - Cambridge GaN Devices Ltd\Documents\CGD documents\Projects\T12012\dynamic'
output_name = 'T12012_vd_overshoot_multipulse_part3.csv'
output_path = os.path.join(output_folder, output_name)
csv_file = open(output_path, mode="w", newline="", encoding="utf-8")
fieldnames = ['File Name', 'chip ID', 'Vd (V)', 'Inductance (H)', 'Vg (V)', 'Rg ON (Ohm)', 'Rg OFF (Ohm)', 'On Time (us)', 'Off Time (us)', 'Temperature (°C)', 'Pulse Number', 'Current (A)', 'Vd Overshoot (V)', 'dIs/dt peak (A/ns)', 'dIs/dt average (A/ns)', 'Vgi max (V)', 'Vgi turn off delay time (ns)']
csv_writer = csv.writer(csv_file)
csv_writer.writerow(fieldnames)

# input_folder = r'\\10.44.21.20\CGDSharedDrive\!USERS\KLS\TV2.3-multipulse\dead-time_550ns_substract'
input_folder = r'\\10.44.21.20\CGDSharedDrive\!USERS\KLS\TV2.3-multipulse\20260804'
files = os.listdir(input_folder)
tss_files = [f for f in files if f.endswith('.tss')]
```

```python
for tss_file in tss_files:
    names = tss_file[:-4].split('_')
    chip_id = names[0]
    vd = int(names[1][:-1])
    induc = eval(names[2][:-2])
    Vg = int(names[4][:-2])
    Rg_on = eval(names[5].split('R')[0])
    Rg_off = eval(names[5].split('R')[1])
    on_time = eval(names[3].split('us')[0])
    off_time = eval(names[3].split('us')[1])
    pulse_num = int(names[6][:-1])
    temperature = 25 if names[-1] == 'RT' else 125
    if vd == 0:
        continue

    data = vd_overshoot_extraction(input_folder, tss_file, pulse_num)
    for line in data:
        csv_writer.writerow([
                tss_file,
                chip_id,
                vd,
                induc,
                Vg,
                Rg_on,
                Rg_off,
                on_time,
                off_time,
                temperature,
                *line
            ])

    print(f"Processed {tss_file}")

csv_file.close()
```
